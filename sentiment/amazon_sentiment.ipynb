```python
example = data['Text'][50]
example
```

```python
from nltk.tokenize import word_tokenize
```

```python
nltk.word_tokenize(example)
```

```python
# Try tokenizing with explicit error handling
try:
  tokens = word_tokenize(example)
  print(tokens[:10])
except LookupError as e:
  print("Error:", e)
  
  # If it still doesn't work, use the simple tokenizer as fallback
  print("\nFalling back to basic tokenization:")
  basic_tokens = example.split()
  print(basic_tokens[:10])
```

```python
tagged = nltk.pos_tag(tokens)
tagged[:10]
```

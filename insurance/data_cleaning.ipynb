```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("./DATA/insurance_data.csv")
data.head()
```

## Cleaning our data

```python
missing_values_row = data.isnull().any(axis=1)
print("Rows with missing values:")
print(data[missing_values_row])
```

```python
# check duplicates
duplicates = data[data.duplicated()]
print("Duplicate rows:")
print(duplicates)
```

```python
# calculatye percentage of each columns
percentage = data.isnull().mean() * 100
print(percentage)
```

```python
data.dropna(axis=0, inplace=True)
data.head()
```

```python
# saving the cleaned data
data.to_csv("./DATA/cleaned_insurance_data.csv", index=False)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
data = pd.read_csv("./DATA/cleaned_insurance_data.csv")
data.head()
```

```python
# calculate the correlation matrix for age
age_claim_corr = data['age'].corr(data['claim'])
print(f"Correlation between age and claim: {age_claim_corr:.2f}")
```

```python
mean_claim = data.groupby('gender')['claim'].mean().reset_index()
mean_claim
```

```python
median_clam = data.groupby('gender')['claim'].median().reset_index()
median_clam
```

```python
# merge mean and median claim data
gender_stats = pd.merge(mean_claim, median_clam, on='gender',suffixes=('_mean', '_median'))
gender_stats
```

```python
# visualize the gender STATS

plt.figure(figsize=(10, 6))
sns.scatterplot(data=data, x='age', y='claim', hue='gender')
plt.title('Scatter Plot of Age vs Claim by Gender')
plt.xlabel('Age')
plt.ylabel('Claim')
plt.show()
```

```python
# bar plot
plt.figure(figsize=(10, 6))
sns.barplot(x='gender', y='claim_mean', data=gender_stats, palette='viridis')
plt.title('Claim by Gender')
plt.xlabel('Gender')
plt.ylabel('Claim')
plt.show()
```

## Finding correlation between BMI and BP with insurance amount

```python
# calculate the correlation matrix between bmi and claim
bmi_claim_corr = data['bmi'].corr(data['claim'])
print(f"Correlation between BMI and claim: {bmi_claim_corr:.2f}")
```

```python
# callculate the correlation matrix between bp and claim
corr_bp_claim = data['bloodpressure'].corr(data['claim'])
print(f"Correlation between BP and claim: {corr_bp_claim:.2f}")
```

```python
# correlation matrix between bp, bmi, and claim
corr_bp_bmi_claim = data[['bloodpressure', 'bmi', 'claim']].corr()
print("Correlation matrix between BP, BMI, and Claim:")
print(corr_bp_bmi_claim)
```

```python
# visualize the data

plt.figure(figsize=(10, 6))
sns.scatterplot(data=data, x='bmi', y='claim', hue='gender')
plt.title('Scatter Plot of BMI vs Claim by Gender')
plt.xlabel('BMI')
plt.ylabel('Claim')
plt.show()
```

```python
# visualizing the data with a subplot
plt.figure(figsize=(15, 8))
plt.subplot(1, 2, 1)
plt.scatter(data['bloodpressure'], data['claim'],alpha=0.5, color='orange')
plt.xlabel('Blood Pressure')
plt.ylabel('Claim')
plt.title('Scatter Plot of Blood Pressure vs Claim')

plt.subplot(1, 2, 2)
plt.scatter(data['bmi'], data['claim'],alpha=0.5, color='blue')
plt.xlabel('BMI') 
plt.ylabel('Claim')
plt.title('Scatter Plot of BMI vs Claim')
plt.tight_layout()
plt.show()
```

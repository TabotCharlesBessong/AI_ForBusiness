```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import LabelEncoder
```

```python
# categorical features
data = pd.read_csv("./DATA/cleaned_insurance_data.csv")
data.head()
```

```python
for col in ['gender','smoker','diabetic','region']:
  le = LabelEncoder()
  data[col] = le.fit_transform(data[col])
```

```python
data.head()
```

```python
features = ['age', 'bmi','gender','bloodpressure','smoker','diabetic','region','children']
X = data[features]
y = data['claim']

model = RandomForestRegressor(random_state=42)
model.fit(X, y)
```

```python
feauture_importances = model.feature_importances_
feature_importances_series = pd.Series(feauture_importances,index=features)
print(feature_importances_series)
```

```python
importances_sorted = feature_importances_series.sort_values(ascending=False)
print(importances_sorted)
```

```python
corr_matrix = data[features + ['claim']].corr()
importance_data = corr_matrix.copy()
importance_data['importance'] = importances_sorted
importance_data.loc['importance'] = pd.concat([feature_importances_series,pd.Series([float('nan')],index=['claim'])])
importance_data['importance'] = pd.concat([feature_importances_series,pd.Series([float('nan')],index=['claim'])])
```

```python
# creating a heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(importance_data, annot=True, cmap='viridis', vmin=0, vmax=1,center=0,linewidths=0.5,linecolor='black')
plt.title('Feature Importance Heatmap')
plt.show()
```

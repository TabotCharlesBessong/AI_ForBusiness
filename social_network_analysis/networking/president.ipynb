```python
import pandas as pd
import networkx as nx
import warnings
```

```python
data = pd.read_csv("../DATA/us_edgelist.csv")
data.head()
```

```python
us_graph = nx.from_pandas_edgelist(data,source="From",target="To")
type(us_graph)
# print(nx.classes.function.info(us_graph))
print(nx.degree_histogram(us_graph))
print("Number of nodes:", us_graph.number_of_nodes())
print("Number of edges:", us_graph.number_of_edges())
print("Is directed:", us_graph.is_directed())
```

```python
print(us_graph.nodes())
print(len(us_graph.nodes()))
```

```python
print(len(us_graph.edges()))
us_graph.add_edge("Franklin","Lincoln")
print(us_graph.nodes())
```

```python
# General connection
nx.degree(us_graph)
```

```python
nx.degree(us_graph,"Obama")
```

```python
nx.degree_centrality(us_graph)
```

```python
sorted(nx.degree_centrality(us_graph).values())
```

```python

most_influential = nx.degree_centrality(us_graph)
for w in sorted(most_influential, key=most_influential.get, reverse=True):
  print(w, most_influential[w])
```

```python
nx.eigenvector_centrality(us_graph)
```

```python
most_important_link = nx.eigenvector_centrality(us_graph)
```

```python
for w in sorted(most_important_link, key=most_important_link.get, reverse=True):
  print(w, most_important_link[w])
```
